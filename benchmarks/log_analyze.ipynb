```python
import re
import statistics
import numpy as np
import matplotlib.pyplot as plt

def analyze_update_world_log(file_path, start_time=10.0, end_time=70.0):
    """
    分析日誌檔案中 update_world 的處理時間
    
    Args:
        file_path: 日誌檔案路徑
        start_time: 取樣開始時間（秒）
        end_time: 取樣結束時間（秒）
    """
    
    # 使用 errors='replace' 將無法解碼的字元替換為 �
    with open(file_path, 'r', encoding='utf-8', errors='replace') as f:
        lines = f.readlines()
    
    # 儲存開始時間的列表
    start_times = []
    processing_times = []
    
    # 正則表達式來提取時間戳記
    pattern = r'\[visualizer ([\d.]+)s\]'
    
    for line in lines:
        match = re.search(pattern, line)
        if not match:
            continue
            
        timestamp = float(match.group(1))
        
        # 只處理指定時間範圍內的資料
        if timestamp < start_time or timestamp > end_time:
            continue
        
        if '開始處理訊息: update_world' in line:
            start_times.append(timestamp)
        elif '完成處理訊息: update_world' in line:
            if start_times:
                start_t = start_times.pop(0)
                duration = timestamp - start_t
                processing_times.append({
                    'start': start_t,
                    'end': timestamp,
                    'duration': duration
                })
    
    # 輸出結果
    print(f"=== Update World 處理時間分析 ===")
    print(f"取樣時間範圍: {start_time}s ~ {end_time}s")
    print(f"找到配對數量: {len(processing_times)}\n")
    
    if processing_times:
        # print("詳細資料:")
        # for i, item in enumerate(processing_times, 1):
        #     print(f"{i:3d}. {item['start']:8.3f}s 開始 → {item['end']:8.3f}s 完成："
        #           f" {item['duration']:.6f}秒 ({item['duration']*1000:.3f}毫秒)")
        
        # 計算統計資料
        durations = [item['duration'] for item in processing_times]
        avg_time = sum(durations) / len(durations)
        max_time = max(durations)
        min_time = min(durations)
        std_time = statistics.stdev(durations) if len(durations) > 1 else 0.0
        
        print(f"\n=== 統計結果 ===")
        print(f"平均處理時間: {avg_time:.6f}秒 ({avg_time*1000:.3f}毫秒)")
        print(f"標準差:       {std_time:.6f}秒 ({std_time*1000:.3f}毫秒)")
        print(f"最長處理時間: {max_time:.6f}秒 ({max_time*1000:.3f}毫秒)")
        print(f"最短處理時間: {min_time:.6f}秒 ({min_time*1000:.3f}毫秒)")
    else:
        print("在指定時間範圍內沒有找到任何配對資料")
    
    return avg_time, std_time
```

```python
num_car_list=[]
avg_time_list=[]
```

```python
# 請將 'your_log_file.txt' 替換成你的日誌檔案路徑
log_file = 'visualizer_other_car0.txt'

# 分析日誌，取樣時間從 10.000s 到 70.000s
results = analyze_update_world_log(log_file, start_time=10.0, end_time=70.0)

num_car_list.append("1")
avg_time_list.append(results[0])
```
